```python
from scipy.signal import square, sawtooth, chirp
```

```python
# x = np.array([5, 2, 6])
# x = np.arange(0, 20, 3)
# x = np.linspace(0, 19, 50
x = np.linspace(0, 1, 10)
y = square(2 * np.pi * 5 * x)

plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.plot(x[-50:], y[-50:], 'o--', label='5 Hz');

plt.subplot(1, 2, 2)
y2 = square(2 * np.pi * 3 * x)
plt.plot(x[-50:], y2[-50:], 'o--', label='3 Hz');
plt.legend();

plt.savefig('chirp.png')
```

```python
x_obs = np.random.normal(0, 3, size=10000)
y_obs = 3 * x_obs ** 2
noise = np.random.normal(0, 10, size=10000)
y_obs = y_obs + noise
# plt.plot(x_obs, y_obs, '.', alpha=.05);
plt.hist2d(x_obs, y_obs, cmap='Blues', bins=200);


x = np.linspace(np.min(x_obs), np.max(x_obs), 1000)
y = 3 * x ** 2
plt.plot(x, y, '--k');

```
