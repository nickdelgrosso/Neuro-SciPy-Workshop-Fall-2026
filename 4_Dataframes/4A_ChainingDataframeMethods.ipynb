# Pandas and DataFrames

Often, we have tables of data--collections of named columns arranged in rows.  The **Pandas** package gives us a **DataFrame()** class that lets us index these columns the same way as with dicts, while still getting the benefit of Numpy arrays, meaning we can still write vectorized code.  

Because Pandas makes it easy to do many analysis steps in a single line, getting used to working with Pandas often means getting used to **Method Chaining**: calling a series of methods all at once.

### Logical Indexing

Logical Masking works just as it does in Numpy, making for a concise way to search for data.

| Code | Description |
| :-- | :-- |
| `df[df['Time'] > 3]` | Get only the rows, for which the "Time" column is greater than 3 |
| `len(df[df['Time'] > 3])` | Count how many rows there are, for which the "Time" column is greater than 3 |

**Example**: How many trials used an angle of 150?

```python
len(df[df['Angle'] == 150])
```

## Group By

Doing the same analysis for different groups of data is enabled with the `.groupby()` method.  This does the analysis **for each** same-valued group in a column.

| Code | Description |
| :-- | :-- |
| `df.groupby('Age').Time.mean()` | For each age in the "Age" column, calculate the mean "Time" value. |
| `df.groupby(['Age', 'Sex']).Time.mean()` | Break down the average time by both age and sex. |

**Example:** What was the response accuracy for matching and non-matching trials?

```python
df.groupby('Matching').IsCorrect.mean()
```
